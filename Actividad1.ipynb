# Ejercicio 4

```python
class Persona:
    """Representa a una persona con nombre y edad."""

    def __init__(self, nombre, edad=0):
        self.nombre = nombre
        self.edad = edad

    def mostrar_edad(self):
        print(f"La edad de {self.nombre} es: {self.edad} años")


class Familia:
    """Calcula las edades de Juan, Alberto, Ana y la mamá."""

    def __init__(self, edad_juan):
        self.juan = Persona("Juan", edad_juan)
        self.alberto = Persona("Alberto")
        self.ana = Persona("Ana")
        self.mama = Persona("La mamá")
        self.calcular_edades()

    def calcular_edades(self):
        self.alberto.edad = (2 / 3) * self.juan.edad
        self.ana.edad = (4 / 3) * self.juan.edad
        self.mama.edad = self.juan.edad + self.alberto.edad + self.ana.edad

    def mostrar_edades(self):
        self.juan.mostrar_edad()
        self.alberto.mostrar_edad()
        self.ana.mostrar_edad()
        self.mama.mostrar_edad()


def main():
    edad_juan = float(input("Ingrese la edad de Juan: "))
    familia = Familia(edad_juan)
    familia.mostrar_edades()


if __name__ == "__main__":
    main()
```

# Ejercicio 5

```python
class SeguimientoVariables:
    def __init__(self):
        self.suma = 0
        self.x = 0
        self.y = 0
        self.traza = []

    def registrar(self, instruccion):
        self.traza.append(
            f"{instruccion:<25} -> SUMA={self.suma}, X={self.x}, Y={self.y}"
        )

    def ejecutar(self):
        self.suma = 0
        self.registrar("SUMA = 0")

        self.x = 20
        self.registrar("X = 20")

        self.suma = self.suma + self.x
        self.registrar("SUMA = SUMA + X")

        self.y = 40
        self.registrar("Y = 40")

        self.x = self.x + self.y ** 2
        self.registrar("X = X + Y ** 2")

        self.suma = self.suma + self.x / self.y
        self.registrar("SUMA = SUMA + X / Y")

    def mostrar_traza(self):
        print("Seguimiento paso a paso:")
        for paso in self.traza:
            print(" ", paso)

    def mostrar_resultado(self):
        print(f"\nEL VALOR DE LA SUMA ES: {self.suma}")


def main():
    algoritmo = SeguimientoVariables()
    algoritmo.ejecutar()
    algoritmo.mostrar_traza()
    algoritmo.mostrar_resultado()


if __name__ == "__main__":
    main()
```

# Ejercicio 12


```python
class Empleado:
    PORCENTAJE_RETENCION = 0.125

    def __init__(self, horas_trabajadas, valor_hora):
        self.horas_trabajadas = horas_trabajadas
        self.valor_hora = valor_hora
        self.salario_bruto = 0
        self.retencion_fuente = 0
        self.salario_neto = 0

    def calcular_salario(self):
        self.salario_bruto = self.horas_trabajadas * self.valor_hora
        self.retencion_fuente = self.salario_bruto * self.PORCENTAJE_RETENCION
        self.salario_neto = self.salario_bruto - self.retencion_fuente

    def mostrar_resultado(self):
        print(f"Salario bruto: ${self.salario_bruto:,.2f}")
        print(f"Retencion en la fuente (12.5%): ${self.retencion_fuente:,.2f}")
        print(f"Salario neto: ${self.salario_neto:,.2f}")


def main():
    horas = float(input("Ingrese las horas trabajadas en la semana: "))
    valor_hora = float(input("Ingrese el valor de la hora: "))

    empleado = Empleado(horas, valor_hora)
    empleado.calcular_salario()
    empleado.mostrar_resultado()


if __name__ == "__main__":
    main()
```

# Ejercicio 14

```python
class Numero:
    def __init__(self, valor):
        self.valor = valor
        self.cuadrado = 0
        self.cubo = 0

    def calcular_potencias(self):
        self.cuadrado = self.valor ** 2
        self.cubo = self.valor ** 3

    def mostrar_resultado(self):
        print(f"Numero: {self.valor}")
        print(f"Cuadrado: {self.cuadrado}")
        print(f"Cubo: {self.cubo}")


def main():
    valor = float(input("Ingrese un numero: "))
    numero = Numero(valor)
    numero.calcular_potencias()
    numero.mostrar_resultado()


if __name__ == "__main__":
    main()
```

# Ejercicio 17


```python
import math


class Circulo:
    def __init__(self, radio):
        self.radio = radio
        self.area = 0
        self.longitud_circunferencia = 0

    def calcular_area(self):
        self.area = math.pi * self.radio ** 2
        return self.area

    def calcular_longitud_circunferencia(self):
        self.longitud_circunferencia = 2 * math.pi * self.radio
        return self.longitud_circunferencia

    def mostrar_resultado(self):
        print(f"Radio: {self.radio}")
        print(f"Area del circulo: {self.area:.2f}")
        print(f"Longitud de la circunferencia: {self.longitud_circunferencia:.2f}")


def main():
    radio = float(input("Ingrese el radio del circulo: "))
    circulo = Circulo(radio)
    circulo.calcular_area()
    circulo.calcular_longitud_circunferencia()
    circulo.mostrar_resultado()


if __name__ == "__main__":
    main()
```
